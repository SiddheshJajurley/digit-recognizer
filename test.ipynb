```python
from sklearn import datasets
from PIL import Image
import numpy as np

digits = datasets.load_digits()

for digit_label in range(10):
    # Find first occurrence of this digit
    idx = np.where(digits.target == digit_label)[0][0]
    digit_image = digits.images[idx]

    img_array = (digit_image / 16.0 * 255).astype(np.uint8)
    img = Image.fromarray(img_array, mode='L')
    img.save(f'digit_{digit_label}.png')

    print(f"Saved digit_{digit_label}.png (label: {digit_label})")
```
